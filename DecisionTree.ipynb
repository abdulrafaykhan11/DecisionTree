```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn import metrics
import seaborn as sns
from sklearn.metrics import accuracy_score
%matplotlib inline

import warnings
warnings.filterwarnings('ignore')
```

```python
path= 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-ML0101EN-SkillsNetwork/labs/Module%203/data/drug200.csv'
my_data = pd.read_csv(path)
my_data
```

```python
my_data.info()
```

```python
label_encoder=LabelEncoder()
```

```python
my_data['Sex'] = label_encoder.fit_transform(my_data['Sex'])
my_data['BP'] = label_encoder.fit_transform(my_data['BP'])
my_data['Cholesterol'] = label_encoder.fit_transform(my_data['Cholesterol'])
my_data
```

```python
my_data.isnull().sum()
```

```python
custom_map = {'drugA':0,'drugB':1,'drugC':2,'drugX':3,'drugY':4}
my_data['Drug_num'] = my_data['Drug'].map(custom_map)
my_data
```

```python
sns.heatmap(my_data.corr(numeric_only=True),annot=True)
```

```python
category_counts = my_data['Drug'].value_counts()

# Plot the count plot
plt.bar(category_counts.index, category_counts.values, color='blue')
plt.xlabel('Drug')
plt.ylabel('Count')
plt.title('Category Distribution')
plt.xticks(rotation=45)  # Rotate labels for better readability if needed
plt.show()
```

```python
X = my_data.drop(['Drug','Drug_num'],axis=1)
y = my_data['Drug_num']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=32)
```

```python
drugTree = DecisionTreeClassifier(criterion='entropy',max_depth=4)
```

```python
drugTree.fit(X_train,y_train)
```

```python
y_pred = drugTree.predict(X_test)
```

```python
accuracy = accuracy_score(y_test,y_pred)
accuracy
```

```python
plot_tree(drugTree,filled=True)
plt.show()
```

```python
drugTree = DecisionTreeClassifier(criterion="entropy", max_depth = 3)
drugTree.fit(X_train,y_train)
tree_predictions = drugTree.predict(X_test)
print("Decision Trees's Accuracy: ", metrics.accuracy_score(y_test, tree_predictions))
```
